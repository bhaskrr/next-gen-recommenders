# Data Preprocessing for Recommender System

## Objective
This notebook prepares the dataset for model training and evaluation.

We will:
- Clean and validate raw interactions
- Transform ratings into implicit feedback (if needed)
- Filter sparse users and items
- Prepare data for chronological splitting
- Export clean interaction data

Important: No statistical learning or feature engineering that uses global data will be performed here to avoid data leakage.

## Imports and Configuration

```python
import pandas as pd
```

```python
class Dataset():
    def __init__(self, path):
        self.path = path
        self.dataset = pd.read_csv(self.path)
```

## Load Raw Dataset

Load the raw interactions dataset.

```python
rating_dataset = Dataset("../data/raw/ml-latest-small/ratings.csv")
```

```python
rating_dataset.dataset.head()
```

```python
rating_dataset.dataset.info()
```

## Initial Data Inspection

We inspect:
- Missing values
- Duplicates
- Basic statistics
- Unique users/items

```python
print("Null Values:", end="\n")
print(rating_dataset.dataset.isnull().sum())
print("Duplicate Values:", rating_dataset.dataset.duplicated().sum())

print("Unique users:", rating_dataset.dataset["userId"].nunique())
print("Unique items:", rating_dataset.dataset["movieId"].nunique())
print("Total interactions:", len(rating_dataset.dataset))
```

## Data Cleaning

- Ensure correct data types

```python
rating_dataset.dataset["timestamp"] = pd.to_datetime(rating_dataset.dataset["timestamp"], unit="s")
rating_dataset.dataset.head()
```
